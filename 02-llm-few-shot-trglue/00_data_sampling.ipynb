# 00 · Data Sampling

Random subsets of [TrGLUE](https://huggingface.co/datasets/turkish-nlp-suite/TrGLUE) used in the prompting experiments:

| Task | Train (few-shot pool) | Test |
|---|---|---|
| SST-2 (sentiment) | 5,000 of 60K | 1,000 of 9K |
| CoLA (grammatical acceptability) | 5,000 of 8K | 1,000 (full) |

```python
import torch
import random
import time
import pandas as pd
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModel
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import matplotlib.pyplot as plt 
from sklearn.manifold import TSNE
import seaborn as sns
import os
```

```python
# Create a directory for checkpoints
checkpoint_dir = "odev2_ha"
os.makedirs(checkpoint_dir, exist_ok=True)
```

```python
# 1. Load the data

# Load the CoLA and SST-2 datasets
ds_cola = load_dataset("turkish-nlp-suite/TrGLUE", "cola")
ds_sst2 = load_dataset("turkish-nlp-suite/TrGLUE", "sst2")

# Create indices for random sampling
indice_train_cola = random.sample(range(len(ds_cola['train'])), 5000)  
indice_train_sst2 = random.sample(range(len(ds_sst2['train'])), 5000)  
indice_test_sst2 = random.sample(range(len(ds_sst2['test'])), 1000)

# CoLA: random sample of the training set (8K to 5K), test set unchanged
cola_train=[ds_cola['train'][i] for i in indice_train_cola]   # 5000 random samples
cola_test = ds_cola['test']  # keep the test set unchanged

# SST-2: random sample of train (60K to 5K) and test (9K to 1K)
sst2_train = [ds_sst2['train'][i] for i in indice_train_sst2]  # 5000 random samples
sst2_test = [ds_sst2['test'][i] for i in indice_test_sst2]  # 1000 random samples
```

```python
# Check results
print("SST-2 train set size:", len(sst2_train))  # should be 5000
print("SST-2 test set size:", len(sst2_test))    # should be 1000
print("CoLA train set size:", len(cola_train))  # should be 5000
print("CoLA test set size:", len(cola_test))    # should stay unchanged
```

```python
# Convert CoLA training data to a pandas DataFrame
cola_train_df = pd.DataFrame(cola_train)
cola_test_df = pd.DataFrame(cola_test)

# Convert SST-2 training data to a pandas DataFrame
sst2_train_df = pd.DataFrame(sst2_train)
sst2_test_df = pd.DataFrame(sst2_test)
```

```python
# Save data as CSV to the given directory
cola_train_df.to_csv(os.path.join(checkpoint_dir, "cola_train.csv"), index=False)  # CoLA train set
cola_test_df.to_csv(os.path.join(checkpoint_dir, "cola_test.csv"), index=False)    # CoLA test set

sst2_train_df.to_csv(os.path.join(checkpoint_dir, "sst2_train.csv"), index=False)  # SST-2 train set
sst2_test_df.to_csv(os.path.join(checkpoint_dir, "sst2_test.csv"), index=False)    # SST-2 test set

# Done
print(f"Data saved to '{checkpoint_dir}' as CSV.")
```
